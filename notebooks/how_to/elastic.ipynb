```python
from geomfum.dataset import NotebooksDataset
from geomfum.operator.elastic import (
    ElasticShellHessian,
    ElasticSpectrumFinder,
    ElasticShellHessianFinder,
)
from geomfum.numerics.eig import ScipyEigsh
from geomfum.shape import TriangleMesh
```

```python
dataset = NotebooksDataset()
mesh = TriangleMesh.from_file(dataset.get_filename("cat-00"))
```

```python
spectrum_finder = ElasticSpectrumFinder()

eigvals, eigvecs = spectrum_finder(mesh, as_basis=False)

(eigvals.shape, eigvecs.shape)
```

```python
import polyscope as ps

ps.init()
ps_mesh = ps.register_surface_mesh("mesh", mesh.vertices, mesh.faces)
ps_mesh.add_scalar_quantity("elastic spectrum", eigvecs[:, 3])
ps.show()
```
